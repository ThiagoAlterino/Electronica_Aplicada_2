## SECCIÓN II: COMPARADOR CON HISTÉRESIS (AD823 VS. LTC6572)

Para el diseño del comparador inversor con histeresis se va a implementar el siguiente circuito:



- Calculo de Tension de umbral inferior



$$ R_2 // R_f = (\frac{1}{R_2} + \frac{1}{R_f})^{-1} = (\frac{R_2 + R_f}{R_2·R_f})^{-1} =  \frac{R_2·R_f}{R_2 + R_f}$$

$$ I_1 = \frac{5V - V_{TL}}{R_1} = \frac{V_{TL}}{\frac{R_2·R_f}{R_2 + R_f}} \quad  \implies \quad \frac{5V}{R_1} - \frac{V_{TL}}{R_1} = \frac{R_2 + R_f}{R_2·R_f} · V_{TL} \quad  \implies \quad \frac{5V}{R_1} = (\frac{R_2 + R_f}{R_2·R_f} + \frac{1}{R_1})·V_{TL} $$

$$ \frac{5V}{R_1} = (\frac{R_1·R_2 + R_1·R_f+R_2·R_f}{R_1·R_2·R_f}) V_{TL} \quad  \implies \quad V_{TL} = \frac{5V}{R_1}·\frac{R_1·R_2·R_f}{R_1·R_2 + R_1·R_f+R_2·R_f} $$

$$ V_{TL} = 5V·\frac{R_2·R_f}{R_1·R_2 + R_1·R_f+R_2·R_f} \quad y \quad V_{TL} = 2V \quad  \implies \quad 2V = 5V·\frac{R_2·R_f}{R_1·R_2 + R_1·R_f+R_2·R_f} $$

$$ \boxed{\frac{R_2·R_f}{R_1·R_2 + R_1·R_f+R_2·R_f} = \frac{2V}{5V} = 0.4} (1) $$

- Calculo de Tension de umbral superior



$$ I_1 = I_2 + I_f \quad y \quad I_1 = \frac{5V - V_{TH}}{R_1} \quad y \quad I_2 = \frac{V_{TH}}{R_2} \quad y \quad I_f = \frac{V_{TH} - V_o}{R_f} = \frac{V_{TH} - 5V}{R_f} $$

$$ \frac{5V}{R_1} - \frac{V_{TH}}{R_1} = \frac{V_{TH}}{R_2} + \frac{V_{TH}}{R_f} - \frac{5V}{R_f} \quad  \implies \quad  (\frac{1}{R_1} + \frac{1}{R_f})·5V = (\frac{1}{R_1} + \frac{1}{R_2} + \frac{1}{R_f})·V_{TH} $$

$$ \frac{R_1 + R_f}{R_1 · R_f}·5V = \frac{R_2·R_f + R_1·R_f + R_1·R_2}{R_1·R_2·R_f}·V_{TH} $$

$$ \frac{R_1 + R_f}{R_1 · R_f}·\frac{R_1·R_2·R_f}{R_2·R_f + R_1·R_f + R_1·R_2}·5V = V_{TH} \quad \implies \quad \frac{R_1·R_2 + R_2·R_f}{R_2·R_f + R_1·R_f + R_1·R_2}·5V = V_{TH} $$

$$ V_{TH} = 2.5V \quad  \implies \quad \frac{R_1·R_2 + R_2·R_f}{R_2·R_f + R_1·R_f + R_1·R_2}·5V = 2.5V $$

$$ \boxed{\frac{R_1·R_2 + R_2·R_f}{R_2·R_f + R_1·R_f + R_1·R_2} = \frac{2.5V}{5V} = 0.5} (2) $$

- Se adopta $\boxed{R_f = 100k\ohm}$ y se divide (2) por (1):

$$ \frac{R_1·R_2 + R_2·R_f}{R_2·R_f + R_1·R_f + R_1·R_2}·\frac{R_1·R_2 + R_1·R_f+R_2·R_f}{R_2·R_f}  = \frac{0.5}{0.4} = 1.25 $$

$$ \frac{R_1·R_2 + R_2·R_f}{R_2·R_f} = 1.25 \quad \implies \quad \frac{R_1}{R_f} + 1 = 1.25 \quad \implies \quad \frac{R_1}{R_f} = 0.25 $$

$$ \boxed{R_1 = 0.25·R_f = 0.25·100k\ohm = 25k\ohm} $$

- De la expresion (1) se obtiene:

$$ \frac{R_2·R_f}{R_1·R_2 + R_1·R_f+R_2·R_f} = 0.4 \quad \implies \quad R_2·R_f = 0.4·(R_1·R_2 + R_1·R_f+R_2·R_f) $$

$$ R_2·R_f = 0.4·R_1·R_2 + 0.4·R_1·R_f + 0.4·R_2·R_f \quad \implies \quad R_2·R_f - 0.4·R_1·R_2 - 0.4·R_2·R_f = 0.4·R_1·R_f $$

$$ (0.6·R_f - 0.4·R_1)·R_2 = 0.4·R_1·R_f \quad \implies \quad R_2 = \frac{0.4·R_1·R_f}{0.6·R_f - 0.4·R_1} $$

$$ R_2 = \frac{0.4·25k\ohm·100k\ohm}{0.6·100k\ohm - 0.4·25k\ohm} = 20k\ohm \quad \implies \quad \boxed{R_2 = 20k\ohm} $$
