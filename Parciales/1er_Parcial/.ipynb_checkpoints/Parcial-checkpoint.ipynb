## Planteo inicial y calculo de las ganancias

Para la implementacion se propone una etapa de preamplificacion para cada canal de audio que luego se sumaran en un sumador inversor. Para los canales de audio mono se plantea usar un operacional en configuracion no inversora y para el canal del microfono balanceado se propone utilizar un operacional en configuracion diferencial.

En cada etapa preamplificadora se calculara la ganancia particular de cada una para que en la salida la señal tenga una amplitud de 3V

Etapas no inversoras:
![noInversor](./archivos/noInversor.png)

La ganancia de la configuracion no inversora es:$\quad A_v = 1 + \frac{R_f}{R_1} = \frac{V_o}{V_i}$

Canal mono 1: $V_p = 40mV \quad \implies \quad A_v = \frac{3V}{40mV} = 75 \quad \implies \quad \frac{R_{f1}}{R_{11}} = 75 - 1 = 74 \quad \implies \quad \boxed{\frac{R_{f1}}{R_{11}} = 74}$

Entonces se adopta $R_{f1} = 82K\ohm \quad y \quad R_{11} = 1.1K\ohm \quad \implies \quad \boxed{A_v = 1 + \frac{R_{f1}}{R_{11}} = 1 + \frac{82K\ohm}{1.1K\ohm} = 75.55}$

Canal mono 2: $V_p = 100mV \quad \implies \quad A_v = \frac{3V}{100mV} = 30 \quad \implies \quad \frac{R_{f2}}{R_{12}} = 30 - 1 = 29 \quad \implies \quad \boxed{\frac{R_{f2}}{R_{12}} = 29}$

Entonces se adopta $R_{f2} = 47K\ohm \quad y \quad R_{12} = 1.6K\ohm \quad \implies \quad \boxed{A_v = 1 + \frac{R_{f2}}{R_{12}} = 1 + \frac{47K\ohm}{1.6K\ohm} = 30.375}$

![amplificador_de_instrumentacion](./archivos/amplificador_de_instrumentacion.png)

La ganancia de la configuracion diferenciadora es: $\quad A_v = \frac{V_0}{(V_2-V_1)} = \frac{R_2}{R_1}$

Microfono balanceado: $V_p = 50 mV \quad \implies \quad A_v = \frac{3V}{50mV} = 60 \quad \implies \quad \boxed{\frac{R_{23}}{R_{13}} = 60}$

Entonces se adopta $R_{23} = 91K\ohm \quad y \quad R_{13} = 1.5K\ohm \quad \implies \quad \boxed{A_v = \frac{R_{23}}{R_{13}} = \frac{91K\ohm}{1.5K\ohm} = 60.66}$

## Circuito a implementar

![Mezclador](./archivos/Mezclador.png)

Para las señales mono se diseña una etapa de preamplificacion con una topologia no inversora con una ganancia especifica para que ambas señales en la salida tengan un nivel de 3V. Esta etapa es implementada con un amplificador operacional.

Para la señal del microfono balanceado se utilizo un amplificacion de instrumentacion discreto diseñado con tres opamps. Se utiliza un amplificador de instrumentacion principalmente dos motivos. La primera razon es que si se utilizara un amplificador operacional en modo diferencial los canales del microfono verian cada uno una impedancia de entrada distinta por lo que genera una ganancia distinta en cada señal antes de la resta, desbalanceando la señal. La segunda razon es que la impedancia de entrada que vean las señales sean grandes, ya que idealmente se busca que toda la tension caiga en el operacional.

Incorporacion de los capacitores en las entradas de preamplificacion:

Si bien la topología diferencial teóricamente rechaza la tensión de modo común de 2.5V por resta matemática, se agregaron capacitores en serie a la entrada del diferenciador para anular completamente la componente continua. Esto resulta imprescindible dada la tolerancia real de los componentes (1%), la cual degrada el Rechazo de Modo Común (CMRR) e introduce el riesgo de saturar la salida debido a la elevada ganancia del canal (A_v = 60). Por buena práctica de diseño en audio, se decidió incorporar también redes RC pasa-altos en las entradas no inversoras para garantizar la eliminación de cualquier desvío de continua (offset) imprevisto y proveer el camino a masa para las corrientes de polarización ($I_b$).

## Eleccion de los amplificadores operacionales

Para la implementacion del mezclador de audio se necesitan 6 amplificadores operacionales, dos para la etapa de preamplificacion de las señales mono, tres para el amplificador de intrumentacion para la etapa de preamplificacion de la señal del microfono balanceado y uno para el sumador. Entonces para esto se va a utilizar los integrados TL072 y TL074. 

El TL072 trae dos amplificadores operacionales, lo cuales seran usados para implementar las etapas preamplificadoras de las señales mono. 

El TL074 trae cuatro amplificadores operacionales, de los cuales tres se usaran para la implementacion del amplificador de instrumentacion discreto y el cuarto se utilizara para la implementacion del sumador.

![TL07x_1](./archivos/TL07x_1.png)
![TL07x_2](./archivos/TL07x_2.png)

### TL072:

1) Producto ganancia - ancho de banda (GBW):
    De estas dos etapas la mas exigente es la etapa de la señal mono 1, dado que la ganancia necesaria es de 75.55 veces.
    Esto implica que el $GBW = 75.55·20KH_z = 1.51MH_z$
    Como el TL072 tiene un valor de GBW tipico de $5.25MH_z$, este operacional cumple olgadamente con el requerimiento.

2) Slew Rate (SR):
    El SR minimo requerido es: $SR \geq 2·\pi·f_{max}·V_{out} = 2·\pi·20KH_z·3V = 0.377 \frac{V}{\micro s}$
    El SR del TL072 es de $20\frac{V}{\micro s}$, con lo cual este integrado cumple ampliamente con el requisito.


### TL074:

1) Producto ganancia - ancho de banda (GBW):
    En la etapa del operacional de instrumentacion se busca que tenga una ganancia de 60 veces, por lo tanto el producto de ganancia de ancho de banda necesario es de: $GBW = 60·20KH_z = 1.2 MH_z$.
    Como el TL074 tiene un valor de GBW tipico de $5.25MH_z$, este operacional cumple olgadamente con el requerimiento.

2) Slew Rate (SR):
    El SR minimo requerido es: $SR \geq 2·\pi·f_{max}·V_{out} = 2·\pi·20KH_z·3V = 0.377 \frac{V}{\micro s}$
    El SR del TL074 es tipicamente de $20\frac{V}{\micro s}$, con lo cual este integrado cumple ampliamente con el requisito.

3) Rechazo al modo comun:
   La señal balanceada del microfono tiene un nivel de continua de 2.5V la cual hay que eliminar en la etapa de preamplificacion. El TL074 tiene un rechazo de modo comun tipicamente de 105 dB, lo cual es suficiente para eliminar la componente de continua.

## Analisis Termico

Para hacer el analisis termico se realiza un circuito electrico equivalente, que representa al sistema termico. En este cicuito equivalente las analogias son las siguientes.

    1) Temperatura [ºC] = Tension [V]
    2) Potencia Disipada [W] = Corriente [A]
    3) Resistencia Termica [ºC/W] = Resistencia Electrica [Ω]

![Circuito_Termico](./archivos/Circuito_Termico.png)
$$T_J = T_a + P_d · R_{Ja} \quad ; \quad P_d = P_Q = \Delta V_S · I_Q \quad y \quad \Delta V_S = V^+ - V^- = 30V$$

Siendo:

1) $T_J$: Temperatura de juntura
2) $T_a$: Temperatura de ambiente ($T_a = 40ºC$)
3) $R_{Ja}$: Resistencia termica juntura-ambiente
4) $P_d$: Potencia disipada
5) $P_Q$: Potencia en reposo
6) $I_Q$: Corriente de reposo

![Features](./archivos/Features.png)

Corriente de reposo por canal: $I_{Q_c} = 940\micro A$


## TL072
![Thermal_TL072](./archivos/Thermal_TL072.png)

El TL072:
    
1) Tiene 2 opamps $\quad \implies \quad I_Q = 940\micro A · 2 = 1.88 mA$
2) $R_{Ja} = 85 \frac{ºC}{W}$

$$P_d = 30V * 1.88mA = 56.4 mW \quad \implies \quad \boxed{P_d = 56.4 mW} $$
$$T_J = 40ºC + 56.4 mW · 85 \frac{ºC}{W} = 40ºC + 4.794ºC \quad \implies \quad \boxed{T_J = 44.794 ºC}$$

## TL074
![Thermal_TL074](./archivos/Thermal_TL074.png)

El TL074:
    
1) Tiene 4 opamps $\quad \implies \quad I_Q = 940\micro A · 4 = 3.76 mA$
2) $R_{Ja} = 80 \frac{ºC}{W}$

$$P_d = 30V * 3.76 mA = 112.8 mW \quad \implies \quad \boxed{P_d = 112.8 mW} $$
$$T_J = 40ºC + 112.8 mW · 80 \frac{ºC}{W} = 40ºC + 9.024ºC \quad \implies \quad \boxed{T_J = 49.024 ºC}$$
